# Task 2 -- Numpy and curve fitting

Fill in the function get_2polynomal_from_3_points(). You are required to use the  
function: np.linalg.solve() for the actual computation, i.e. the task is mainly   
about shaping the input so it fits the np.linalg.solve() function. Note: dont use polyfit().

The function should return a 1d numpy array, with shape = (3,) in the format a, b,c.  
Such that ax^2 + bx + c = y, goes through the points p0, p1, p2

HINT: How should a matrix equation be formed, so solving it gives us the coefficients for a polynomial?

Limitations:
* There should be NO loops or ifs! (for/while/if). 
* Numpy functions that might come handy: reshape, hstack, flatten, ones



```python
import numpy as np

def get_2polynomal_from_3_points(p0: np.array, p1: np.array, p2: np.array) -> np.array:
    # -- CODE --
    points = np.vstack((p0, p1, p2))
    print ("points:", points)
    x = points[:, 0]
    y = points[:, 1]
    # print("x:", x)
    # print("y:", y)
   
    col_x2 = (x**2).reshape(3, 1)   # [[x0²], [x1²], [x2²]]
    col_x  = x.reshape(3, 1)        # [[x0],  [x1],  [x2]]
    col_1  = np.ones((3, 1))        # [[1],   [1],   [1]]

    A = np.hstack((col_x2, col_x, col_1))  # [[x0², x0, 1], [x1², x1, 1], [x2², x2, 1]]
    print("A:", A)

    coefficients = np.linalg.solve(A, y)  # Solve for [a, b, c] in Ax = y
    return coefficients

    
point0 = np.array([0.147, 0.596])
point1 = np.array([0.7, 0.992])
point2 = np.array([2.06, 0.17])

polynomial_coefficients = get_2polynomal_from_3_points(point0, point1, point2)
assert(polynomial_coefficients.shape[0] == 3 and polynomial_coefficients.ndim == 1)

print("Polynomial: {:2f}x^2 + {:.2f}x + {:.2f}".format(*polynomial_coefficients))


if not np.isclose(polynomial_coefficients[0], -0.69, atol=0.01):
    print("x^2 coeff is wrong (a), should be close to -0.69")
    
if not np.isclose(polynomial_coefficients[1], 1.3, atol=0.01):
    print("x coeff is wrong (b), should be close to 1.3")
    
if not np.isclose(polynomial_coefficients[2], 0.42, atol=0.01):
    print("Constant factor (c) is wrong, should be close to 0.42")
    
if np.isclose(polynomial_coefficients, [-0.69, 1.3, 0.42], atol=0.01).all():
    print("Correct polynomial found.")
```

```python
%%time
# BASELINE: measure the regular Fortuna (average best loss over many runs)
# Uses predict, sample_theta, get_loss, xs, ys from the cell above -> run that cell first.

import numpy as np


def fortuna(xs, ys, n_params, n_guesses=100000, predict_fn=predict):
    # Regular Fortuna: n_guesses random guesses, keep the best one
    # predict_fn lets us reuse this function for another model (h(x) later)
    best_loss = float('inf')
    best_theta = None

    for _ in range(n_guesses):
        curr_theta = sample_theta(n_params)
        y_hat = predict_fn(xs, curr_theta)
        curr_loss = get_loss(y_hat, ys)

        if curr_loss < best_loss:
            best_loss = curr_loss
            best_theta = curr_theta

    return best_loss, best_theta


n_runs = 100   

losses = []
for run in range(n_runs):
    loss, theta = fortuna(xs, ys, n_params=9)
    losses.append(loss)
    print(f"run {run+1}/{n_runs}: best loss = {loss:.2f}")

print("-" * 40)
print("average best loss:", np.mean(losses))
print("min / max:", np.min(losses), "/", np.max(losses))
```

```python
%%time
# IMPROVED FORTUNA: search around the best theta, with a step size that shrinks over time.
# Big steps at the start (explore), tiny steps at the end (fine-tune).
# We assume that the best theta is somewhere in the neighborhood of the best theta found so far
import numpy as np


def fortuna_shrinking(xs, ys, n_params, n_guesses=100000, step_start=1.0, step_end=0.01, predict_fn=predict):
    # start from one random guess
    best_theta = sample_theta(n_params)
    best_loss = get_loss(predict_fn(xs, best_theta), ys)

    for i in range(n_guesses):
        # step goes linearly from step_start to step_end
        step = step_start + (step_end - step_start) * i / n_guesses

        # draws 9 random numbers centered on 0. Most of them fall between -step and +step
        curr_theta = best_theta + np.random.normal(0, step, size=n_params)
        # keep every parameter inside [-4, 4], same range as sample_theta
        curr_theta = np.clip(curr_theta, -4, 4)
        y_hat = predict_fn(xs, curr_theta)
        curr_loss = get_loss(y_hat, ys)

        if curr_loss < best_loss:
            best_loss = curr_loss
            best_theta = curr_theta

    return best_loss, best_theta


n_runs = 100

losses_shrinking = []
for run in range(n_runs):
    loss, theta = fortuna_shrinking(xs, ys, n_params=9)
    losses_shrinking.append(loss)
    print(f"run {run+1}/{n_runs}: best loss = {loss:.2f}")

print("-" * 40)
print("average best loss:", np.mean(losses_shrinking))
print("min / max:", np.min(losses_shrinking), "/", np.max(losses_shrinking))

# plot the last solution
fig = px.line(x=xs, y=ys, title="f(x) vs improved Fortuna solution")
fig.add_scatter(x=xs, y=predict(xs, theta), mode='lines', name="y_hat")
fig.update_layout(xaxis_range=[xs.min(), xs.max()], yaxis_range=[-6, 6])
fig.show()
```

### Part 2 -- What was changed in Fortuna, why, and how it was measured

**What:** instead of drawing every θ uniformly in $[-4, 4]^9$, the improved version is a *local search* around the best θ found so far:
1. Start from one random θ (drawn with `sample_theta`, unchanged).
2. New candidate = `best_theta + N(0, step)`: small Gaussian noise added to every parameter.
3. The candidate is clipped to $[-4, 4]$ so every parameter stays in the allowed range.
4. Keep the candidate only if its loss is lower.
5. `step` shrinks linearly from `1.0` to `0.01` during the 100 000 guesses.

**Why:**
* Regular Fortuna never uses what it has already learned: every guess is independent. In 9 dimensions the space is huge, so hitting a good θ by pure luck is very unlikely.
* The loss is a *continuous* function of θ: if θ is good, a θ close to it is probably good too. Searching near the best θ exploits that.
* Large steps at the start let the search jump between regions (exploration); small steps at the end fine-tune the solution (exploitation).
* Clipping: without it, the search can drift outside $[-4, 4]$. In our tests the unclipped version sometimes got lost (one run ended with loss 123), clipping removed those outliers.

**How it was measured:** both algorithms get the same budget (100 000 guesses) and are run **100 times**; we compare the average, min and max of the best loss (cells above).

| Algorithm (100 runs, 100 000 guesses) | average best loss | min | max |
|---|---|---|---|
| Regular Fortuna | 40.63 | 14.06 | 64.28 |
| Improved, no clipping | 4.57 | 0.02 | 123.64 |
| **Improved, with clipping** | **1.94** | 0.03 | 6.77 |

The improved version is about **20x better** on average and below the "excellent" target of 5.

```python
%%time
# PART 3: optimize h(x) = mu - (zeta * sin(kappa * x)) * (tau * (x + lambda))
# Uses sample_theta, get_loss, fortuna, fortuna_shrinking, xs, ys_h_1, ys_h_2 from the cells above
# -> run cells 17, 18 (only the def is needed) and 19 first.
# Do not run Task 5 before this cell: it redefines predict and xs.
import numpy as np
import plotly.express as px


def predict_h(x, theta):
    # 5 parameters: mu (vertical offset), zeta and tau (amplitude), kappa (frequency), lam (shift)
    mu, zeta, kappa, tau, lam = theta
    return mu - (zeta * np.sin(kappa * x)) * (tau * (x + lam))


def summary(losses):
    # the median is shown too: a few runs stuck in a local minimum can blow up the average
    return (f"average {np.mean(losses):.2f} | median {np.median(losses):.2f} "
            f"| min {np.min(losses):.2f} | max {np.max(losses):.2f}")


n_params_h = 5
n_runs = 100

# ys_h_2 is written with double brackets -> shape (1, 65), flatten it to (65,)
datasets = {"ys_h_1": ys_h_1.flatten(), "ys_h_2": ys_h_2.flatten()}

for name, ys_target in datasets.items():
    losses_base, losses_impr = [], []
    best_base, best_impr = (float('inf'), None), (float('inf'), None)

    for run in range(n_runs):
        loss_b, theta_b = fortuna(xs, ys_target, n_params_h, predict_fn=predict_h)
        loss_i, theta_i = fortuna_shrinking(xs, ys_target, n_params_h, predict_fn=predict_h)
        losses_base.append(loss_b)
        losses_impr.append(loss_i)
        best_base = min(best_base, (loss_b, theta_b), key=lambda t: t[0])
        best_impr = min(best_impr, (loss_i, theta_i), key=lambda t: t[0])

    print("=" * 50)
    print(name)
    print("  regular Fortuna :", summary(losses_base))
    print("  improved Fortuna:", summary(losses_impr))
    print("  best theta (improved) [mu, zeta, kappa, tau, lam]:", np.round(best_impr[1], 3))

    fig = px.line(x=xs, y=ys_target, title=f"{name} vs Fortuna solutions (best of {n_runs} runs)")
    fig.add_scatter(x=xs, y=predict_h(xs, best_base[1]), mode='lines', name="regular Fortuna")
    fig.add_scatter(x=xs, y=predict_h(xs, best_impr[1]), mode='lines', name="improved Fortuna")
    fig.show()
```

If the random start is in the wrong "valley", the search never leaves it, especially once the step becomes small.
  * Regular Fortuna does not get stuck (every guess is independent), but it is never precise either: it never gets below 25.
  * Note: $\zeta$ and $\tau$ are redundant, only the product $\zeta\tau$ matters. So there are infinitely many equally good θ.
* **`ys_h_2`:** the data jumps abruptly every 5 points (e.g. 5.48 → 16.58, 15.54 → 4.15). $h(x)$ is smooth and cannot reproduce these jumps, so **no θ gets a low loss**: the best possible is around 2833. The improved version reaches that floor in most runs (median 2833.66, a result regular Fortuna never reaches), but it is again worse on average because of the runs stuck in local minima. The gap between the two algorithms is much smaller here, because even the best θ is a poor fit.

**Conclusion:** local search assumes that "close to a good θ = good θ", with one valley to go down into. That holds for the sum of sines, but $h(x)$ has several valleys, so the improved Fortuna becomes a gamble: very good or very bad. A natural fix would be **restarts** (e.g. 10 local searches of 10 000 guesses each from different random starting points, keeping the best one), which combines the exploration of regular Fortuna with the precision of local search.

# Task 5 -- Fortuna For Decision Trees

Implement a Decision Tree for solving the XOR problem.   
Here: there are 2 real valued inputs, x0, x1  (found in xs).  

And the DT should take these two as input and predict an output: 0 or 1.   
The true answer can be found in ys_true.


```python
import numpy as np
rng = np.random.default_rng(42)
n_examples = 40

# 40 points ; chaque point = 2 nombres réels dans [0,1] : x0 (colonne 0) et x1 (colonne 1).
xs = rng.uniform(size=(n_examples, 2))

# make a true y
# "grand" = > 0.5. ys_true = XOR : 1 si EXACTEMENT un des deux est grand, sinon 0.
b = (xs>0.5).astype(int)
ys_true = np.logical_xor(b[:, 0], b[:, 1]).astype(int)


##################################
# here you implement a Decision Tree that is built using the fortuna algorithm.
# The depth of the tree should be less than 4
# The DT should reach 100% accuracy.

# -- CODE --

# BUT : trouver un arbre de décision (profondeur 2) qui reproduit XOR sur les 40 points.
# Fortuna = on tire des arbres au hasard et on garde le meilleur (meilleur score).
frt = np.random.default_rng(0)   # générateur aléatoire pour Fortuna (séparé des données)

# Un arbre est fait de nœuds-questions et de feuilles-réponses :
#   - nœud    (= une liste)  : [variable, seuil, branche_vrai, branche_faux]
#       variable : 0 -> x0, 1 -> x1 ;  test : x[variable] >= seuil
#   - feuille (= un nombre)  : 0 ou 1, la réponse finale.


def grow(depth):
    # Construit UN arbre au hasard. Seules les questions sont tirées au hasard ;
    # les feuilles restent vides (None) et seront remplies plus tard par fit().

    # Cas de base : on a atteint la profondeur maximale -> on crée une feuille vide.
    if depth == 0:
        return None

    # Sinon, on crée un nœud-question avec des valeurs tirées au hasard.
    variable = frt.integers(0, 2)      # variable à tester : 0 ou 1 (x0 ou x1)
    seuil = frt.uniform(0, 1)          # valeur de comparaison (entre 0 et 1)

    # On construit les deux sous-arbres, un cran plus bas.
    branche_vrai = grow(depth - 1)     # utilisé si x[variable] >= seuil
    branche_faux = grow(depth - 1)     # utilisé sinon

    return [variable, seuil, branche_vrai, branche_faux]


def fit(arbre, indices):
    # Remplit les feuilles vides par un VOTE MAJORITAIRE :
    # dans chaque feuille, on écrit la vraie réponse (0 ou 1) la plus fréquente
    # parmi les points de "indices" qui arrivent dans cette feuille.

    # Cas de base : on est sur une feuille vide -> on lui donne sa réponse.
    if arbre is None:
        if len(indices) == 0:
            return 0                                     # aucun point -> 0 par défaut
        reponses = ys_true[indices]                      # vraies réponses des points reçus
        reponse_majoritaire = int(reponses.mean() >= 0.5)  # 1 si plus de 1 que de 0
        return reponse_majoritaire

    # Sinon, on est sur un nœud-question -> on répartit les points dans les 2 branches.
    variable, seuil, branche_vrai, branche_faux = arbre
    part_dans_vrai = xs[indices, variable] >= seuil      # masque : qui va dans "vrai"
    indices_vrai = indices[part_dans_vrai]                # points qui vont dans "vrai"
    indices_faux = indices[~part_dans_vrai]               # points qui vont dans "faux"

    # On remplit chaque sous-arbre avec seulement ses points.
    nouveau_vrai = fit(branche_vrai, indices_vrai)
    nouveau_faux = fit(branche_faux, indices_faux)

    return [variable, seuil, nouveau_vrai, nouveau_faux]


def predict(arbre, X):
    # Fait descendre chaque point dans l'arbre jusqu'à une feuille,
    # puis renvoie la réponse lue dans cette feuille (0 ou 1).
    predictions = []

    for point in X:
        noeud = arbre                                    # on démarre à la racine

        # On descend tant qu'on est sur une question (représentée par une liste).
        while isinstance(noeud, list):
            variable, seuil, branche_vrai, branche_faux = noeud
            valeur = point[variable]

            if valeur >= seuil:
                noeud = branche_vrai                     # on va dans la branche "vrai"
            else:
                noeud = branche_faux                     # on va dans la branche "faux"

        # Ici, noeud est une feuille : un simple nombre (0 ou 1).
        predictions.append(noeud)

    return np.array(predictions)


# --- Fortuna : essayer des arbres au hasard et garder le meilleur ---
meilleur_arbre = None
meilleur_score = -1                                      # -1 : pire score possible au départ

for tentative in range(100000):
    arbre_aleatoire = grow(2)                            # 1) tirer un arbre au hasard
    arbre_complet = fit(arbre_aleatoire, np.arange(len(xs)))  # 2) remplir ses feuilles
    predictions = predict(arbre_complet, xs)             # 3) prédire les 40 points
    score = (predictions == ys_true).mean()              # 4) score = bonnes réponses / 40

    if score > meilleur_score:                           # 5) meilleur que le record ?
        meilleur_arbre = arbre_complet
        meilleur_score = score
        if score == 1.0:                                 # 6) parfait -> on s'arrête
            break

print("accuracy:", meilleur_score, "| profondeur: 2 (< 4)")
print("tree:", meilleur_arbre)
```
